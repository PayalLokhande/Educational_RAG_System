```python
# Force-install libraries natively inside your notebook kernel environment
# Force-install directly inside the active Python 3.14 kernel
%pip install pandas openpyxl

import pandas as pd
import json
import os

print("📊 Initializing Data Diversity & Quality Evaluation Suite...")

# The "../" at the front tells Python to look OUTSIDE the current folder
TECH_JSON = "../data/JSON data files/cleaned_technical_questions.json"
VIVA_JSON = "../data/JSON data files/cleaned_viva_questions.json"

```

```python
print("=== 🛠️ TECHNICAL INTERVIEW TRACK AUDIT ===")
with open(TECH_JSON, 'r', encoding='utf-8') as f:
    tech_data = json.load(f)
df_tech = pd.DataFrame(tech_data)

# 🚀 FORCE STANDARDIZE THE TOPICS DIRECTLY IN YOUR ANALYTICS BLOCK
def force_standardize(topic_name):
    t = str(topic_name).lower().strip()
    if any(x in t for x in ['python', 'lambda', 'exception', 'mutable', 'tuple', 'list', 'dict', 'function']):
        return 'Python'
    elif any(x in t for x in ['oop', 'pattern', 'class', 'interface', 'polymorphism', 'encapsulation', 'abstract']):
        return 'OOPs'
    elif any(x in t for x in ['sql', 'dbms', 'database', 'join', 'acid', 'clause', 'where', 'table', 'query']):
        return 'DBMS'
    elif any(x in t for x in ['web', 'cors', 'http', 'api', 'server', 'socket', 'auth', 'ajax', 'url', 'client']):
        return 'Web Fundamentals'
    else:
        return 'DSA' # Combines sorting, searching, arrays, stacks, queues, big o

# Apply the clean transformation to the column
df_tech['Topic'] = df_tech['Topic'].apply(force_standardize)

print(f"Total Verified Technical Questions: {len(df_tech)}\n")
print("1. Difficulty Level Distribution:")
print(df_tech['Difficulty'].value_counts())

print("\n2. Topic/Subject Domain Distribution:")
print(df_tech['Topic'].value_counts())

print("\n3. Cross-Tabulation Matrix:")
display(pd.crosstab(df_tech['Topic'], df_tech['Difficulty']))
```

```python
print("=== 📋 PROJECT VIVA TRACK AUDIT ===")
with open(VIVA_JSON, 'r', encoding='utf-8') as f:
    viva_data = json.load(f)
df_viva = pd.DataFrame(viva_data)

print(f"Total Verified Project Viva Questions: {len(df_viva)}\n")
print("1. Difficulty Level Distribution:")
print(df_viva['Difficulty'].value_counts())
print("\n2. Core Evaluation Theme Distribution:")
print(df_viva['Topic'].value_counts())
print("\n3. Cross-Tabulation Matrix:")
print(pd.crosstab(df_viva['Topic'], df_viva['Difficulty']))
```

```python
print("=== 🎛️ AI GRADING ENGINE INTEGRITY CHECK ===")

# Check Technical file key name dynamically
tech_key = 'Model Answer' if 'Model Answer' in df_tech.columns else 'model_answer'
if tech_key not in df_tech.columns:
    # Backup check for any case variations
    tech_key = [col for col in df_tech.columns if 'answer' in col.lower()][0]

# Check Viva file key name dynamically
viva_key = 'Model Answer' if 'Model Answer' in df_viva.columns else 'model_answer'
if viva_key not in df_viva.columns:
    viva_key = [col for col in df_viva.columns if 'answer' in col.lower()][0]

# Calculate averages using the detected keys
avg_tech_words = df_tech[tech_key].str.split().str.len().mean()
avg_viva_words = df_viva[viva_key].str.split().str.len().mean()

print(f"Detected Tech Answer Key: '{tech_key}' | Detected Viva Answer Key: '{viva_key}'")
print(f"Average Technical Rubric Length: {avg_tech_words:.1f} words")
print(f"Average Project Viva Rubric Length: {avg_viva_words:.1f} words")
print("\n🏆 AUDIT SUITE COMPLETE: DATASETS OPTIMIZED FOR PRODUCTION!")
```
