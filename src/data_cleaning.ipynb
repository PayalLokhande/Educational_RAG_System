```python
import pandas as pd

viva_df = pd.read_excel("../data/Excel data files/raw_project_viva.xlsx", engine="openpyxl")

print(viva_df.head())
```

```python
# =========================================================
# CELL 9: POSITIONAL INDEPENDENT PIPELINE ENGINE
# =========================================================

# 🚀 FORCE OVERRIDE HEADERS BY STRIP-POSITION (Bypasses all invisible string/space bugs)
viva_df.columns = ["ID", "Topic", "Difficulty", "Question", "Model Answer", "Keywords"]
tech_df.columns = ["ID", "Topic", "Difficulty", "Question", "Model Answer", "Keywords"]

# --- 1. CLEAN TRACK 1: PROJECT VIVA ---
viva_df["Question"] = viva_df["Question"].astype(str).str.strip()
viva_df = viva_df.dropna(subset=["Question"])
viva_df = viva_df[viva_df["Question"] != ""]
viva_df = viva_df.drop_duplicates(subset=["Question"])

# --- 2. CLEAN TRACK 2: TECHNICAL INTERVIEW ---
tech_df["Question"] = tech_df["Question"].astype(str).str.strip()
tech_df = tech_df.dropna(subset=["Question"])
tech_df = tech_df[tech_df["Question"] != ""]
tech_df = tech_df.drop_duplicates(subset=["Question"])

# --- 3. REFACTORING LAYER: TOPIC STANDARDIZATION ---
def standardize_topics(topic_name):
    t = str(topic_name).lower().strip()
    if any(x in t for x in ['python', 'lambda', 'exception', 'mutable', 'tuple', 'list', 'dict', 'function']):
        return 'Python'
    elif any(x in t for x in ['oop', 'pattern', 'class', 'interface', 'polymorphism', 'encapsulation', 'abstract']):
        return 'OOPs'
    elif any(x in t for x in ['sql', 'dbms', 'database', 'join', 'acid', 'clause', 'where', 'table', 'query']):
        return 'DBMS'
    elif any(x in t for x in ['web', 'cors', 'http', 'api', 'server', 'socket', 'auth', 'ajax', 'url', 'client']):
        return 'Web Fundamentals'
    else:
        return 'DSA' # Falls back to Data Structures for standard entries

# Standardize the technical topics column cleanly
tech_df['Topic'] = tech_df['Topic'].apply(standardize_topics)

# --- 4. EXPORT SPRINT: SAVE PRODUCTION JSON FILES ---
viva_df.to_json(
    "../data/JSON data files/cleaned_viva_questions.json",
    orient="records",
    indent=2
)
tech_df.to_json(
    "../data/JSON data files/cleaned_technical_questions.json",
    orient="records",
    indent=2
)

# --- 5. DATABASE SPRINT: SAVE FLAT CSV DATABASE LOAD TABLES ---
tech_df.to_csv("../data/Database Load Files/technical_db_load.csv", index=False)
viva_df.to_csv("../data/Database Load Files/viva_db_load.csv", index=False)

print(f"🎉 SUCCESS! Cleaned Project Viva: {len(viva_df)} rows")
print(f"🎉 SUCCESS! Cleaned Technical Interview: {len(tech_df)} rows")
print("All four JSON and CSV database load files successfully updated inside your repository subfolders!")
```
